```python
#Importar bibliotecas
import requests
from bs4 import BeautifulSoup
import re
```

```python
def scrap_belezanectar_products(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36"
    }
    response = requests.get(url, headers=headers, timeout=100)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')

        valid_links = set()

        for product_div in soup.find_all('div', class_="card card--standard card--media"):
            a = product_div.find("a", href=True)
            text = a.text
            price_span = product_div.find("span", class_="price-item--sale")
            price = price_span.text if price_span else None
            if "esfoliante" in text.lower() and price: print(f"O produto {text.strip()} está custando {price.strip()} na promoção.")

        return sorted(valid_links)
    else:
        return f"Falha ao extrair os produtos da página de produtos: {response.status_code}"
```

```python
#Definindo a função que retornará os links

def scrape_cnn_links(url):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0 Safari/537.36"
    }
    response = requests.get(url, headers=headers, timeout=20)

    # Se a request tiver retorno com sucesso
    if response.status_code == 200:
        # Parsing do conteúdo da página
        soup = BeautifulSoup(response.text, 'html.parser')

        valid_links = set()

        # Encontrar os links de notícias
        #Ele sabe isto aqui porque inspecionou antes o site da CNN
        for a in soup.find_all('a', href=True):
            href = a['href']

            if href.startswith('http'):
                full_link = href
            elif href.startswith('/'):
                full_link = f"https://edition.cnn.com{href}"
            else:
                continue

            # Filtrar links que parecem ser artigos (padr?o com data no caminho)
            if re.search(r"/\d{4}/\d{2}/\d{2}/", full_link):
                full_link = full_link.split('#')[0].split('?')[0]
                valid_links.add(full_link)

        return sorted(valid_links)
    else:
        return f"Falha ao extrair o conte?do da not?cia: {response.status_code}"
```

```python
# Definindo a URL padr?o
url = "https://edition.cnn.com/world"

# Chama a fun??o de scrap dos dados e salva as URLs em um arquivo
links = scrape_cnn_links(url)
with open('CNN_Links.txt', 'w') as file:
    for link in links:
        file.write(link + '\n')

print(f"Total de links coletados: {len(links)}")
```
